```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Number of states:", df["State"].nunique())

print("\nStates:")
print(sorted(df["State"].dropna().unique()))
```

```python
print("Number of districts:", df["District"].nunique())
```

```python
print("Number of crops:", df["Crop"].nunique())

print("\nCrops:")
print(sorted(df["Crop"].dropna().unique()))
```

```python
print("Minimum year:", df["Crop_Year"].min())
print("Maximum year:", df["Crop_Year"].max())
```

```python
year_counts = df["Crop_Year"].value_counts().sort_index()

print(year_counts)
```

```python
print(df["Season"].value_counts())
```

```python
print("Negative Area:",
      (df["Area"] < 0).sum())

print("Negative Production:",
      (df["Production"] < 0).sum())

print("Negative Yield:",
      (df["Yield"] < 0).sum())
```

```python
print("Area = 0:",
      (df["Area"] == 0).sum())

print("Production = 0:",
      (df["Production"] == 0).sum())

print("Yield = 0:",
      (df["Yield"] == 0).sum())
```
