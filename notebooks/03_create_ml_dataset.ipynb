```python
import pandas as pd
import numpy as np
import os

df = pd.read_csv("../data/processed/agriculture_clean.csv")

print("Shape:", df.shape)
df.head()
```

```python
ml_df = df[
    (df["Area"] > 0) &
    (df["Production"] > 0) &
    (df["Yield"] > 0)
].copy()

print("Original cleaned rows:", len(df))
print("ML dataset rows:", len(ml_df))
print("Rows removed:", len(df) - len(ml_df))
```

```python
before_2020_filter = len(ml_df)

ml_df = ml_df[
    ml_df["Crop_Year"] < 2020
].copy()

print("Removed 2020 rows:", before_2020_filter - len(ml_df))
print("Final ML dataset:", len(ml_df))
```

```python
print("Years:")
print(
    ml_df["Crop_Year"]
    .value_counts()
    .sort_index()
)
```

```python
print("States:", ml_df["State"].nunique())
print("Districts:", ml_df["District"].nunique())
print("Crops:", ml_df["Crop"].nunique())
```

```python
print(ml_df.isnull().sum())
```

```python
print(ml_df["Yield"].describe())
```

```python
os.makedirs("../data/processed", exist_ok=True)

ml_df.to_csv(
    "../data/processed/ml_training_base.csv",
    index=False
)

print("ML training base saved!")
print("Shape:", ml_df.shape)
```
