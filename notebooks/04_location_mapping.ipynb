```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv("../data/processed/ml_training_base.csv")

print("Rows:", len(df))
print("Columns:", df.shape[1])
```

```python
locations = (
    df[["State", "District"]]
    .drop_duplicates()
    .sort_values(["State", "District"])
    .reset_index(drop=True)
)

print("Unique State-District combinations:", len(locations))
locations.head(20)
```

```python
district_counts = (
    locations.groupby("State")["District"]
    .nunique()
    .sort_values(ascending=False)
)

district_counts
```

```python
print("Total states:", locations["State"].nunique())
print("Total districts:", locations["District"].nunique())
```

```python
for state in sorted(locations["State"].unique()):
    print("\n" + "=" * 60)
    print(state)
    print("=" * 60)
    print(
        sorted(
            locations.loc[
                locations["State"] == state,
                "District"
            ].dropna().unique()
        )
    )
```

```python
district_data = (
    df.groupby(["State", "District"])
      .agg(
          records=("Yield", "size"),
          crops=("Crop", "nunique"),
          years=("Crop_Year", "nunique")
      )
      .reset_index()
)

district_data.head()
```

```python
location_master = locations.copy()

location_master["Latitude"] = np.nan
location_master["Longitude"] = np.nan

location_master.head()
```

```python
os.makedirs("../data/processed", exist_ok=True)

location_master.to_csv(
    "../data/processed/location_master.csv",
    index=False
)

print("Location master created!")
print("Rows:", len(location_master))
```
