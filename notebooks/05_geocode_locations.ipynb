```python
import pandas as pd
import numpy as np
import time
import os

from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
```

```python
locations = pd.read_csv(
    "../data/processed/location_master.csv"
)

print("Locations:", len(locations))
locations.head()
```

```python
geolocator = Nominatim(
    user_agent="smart-crop-ai-college-project"
)

geocode = RateLimiter(
    geolocator.geocode,
    min_delay_seconds=1
)
```

```python
test_locations = locations.head(5).copy()

for _, row in test_locations.iterrows():

    query = f"{row['District']}, {row['State']}, India"

    location = geocode(query)

    if location:
        print(
            row["State"],
            "|",
            row["District"],
            "→",
            location.latitude,
            location.longitude
        )
    else:
        print(
            row["State"],
            "|",
            row["District"],
            "→ NOT FOUND"
        )
```

```python
def try_geocode(district, state):
    queries = [
        f"{district}, {state}, India",
        f"{district}, India",
    ]

    for query in queries:
        try:
            location = geocode(query)

            if location:
                return {
                    "latitude": location.latitude,
                    "longitude": location.longitude,
                    "matched_address": location.address,
                    "query_used": query
                }

        except Exception as e:
            print("Geocoding error:", e)

    return None
```

```python
test_locations = locations.head(5).copy()

for _, row in test_locations.iterrows():

    result = try_geocode(
        row["District"],
        row["State"]
    )

    if result:
        print(
            row["State"],
            "|",
            row["District"],
            "→",
            result
        )
    else:
        print(
            row["State"],
            "|",
            row["District"],
            "→ NOT FOUND"
        )
```
