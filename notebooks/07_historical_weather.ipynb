```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

geolocator = Nominatim(
    user_agent="smart-crop-ai-college-project"
)

geocode = RateLimiter(
    geolocator.geocode,
    min_delay_seconds=1
)
```

```python
def geocode_final_location(state, district):

    queries = [
        f"{district}, {state}, India",
        f"{district} district, {state}, India"
    ]

    for query in queries:

        try:
            result = geocode(query)

            if result:
                return {
                    "Latitude": result.latitude,
                    "Longitude": result.longitude,
                    "Geocoded_Address": result.address,
                    "Status": "FOUND"
                }

        except Exception as e:
            print("Error:", e)

    return {
        "Latitude": np.nan,
        "Longitude": np.nan,
        "Geocoded_Address": "",
        "Status": "NOT_FOUND"
    }
```

```python
def fetch_weather(lat, lon, year):

    url = "https://archive-api.open-meteo.com/v1/archive"

    params = {
        "latitude": lat,
        "longitude": lon,
        "start_date": f"{year}-01-01",
        "end_date": f"{year}-12-31",
        "daily": ",".join([
            "temperature_2m_mean",
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_sum"
        ]),
        "timezone": "auto",
        "models": "era5"
    }

    for attempt in range(3):

        try:
            response = requests.get(
                url,
                params=params,
                timeout=60
            )

            if response.status_code == 200:
                data = response.json()

                weather = pd.DataFrame(
                    data["daily"]
                )

                weather["latitude"] = lat
                weather["longitude"] = lon
                weather["year"] = year

                return weather

            print(
                f"Request failed: {response.status_code}"
            )

        except Exception as e:
            print(
                f"Attempt {attempt + 1} failed:",
                e
            )

            time.sleep(5)

    return None
```

```python
def download_weather_year(lat, lon, year, output_file):

    # Don't download again if already completed
    if os.path.exists(output_file):
        return "EXISTS"

    url = "https://archive-api.open-meteo.com/v1/archive"

    params = {
        "latitude": lat,
        "longitude": lon,
        "start_date": f"{year}-01-01",
        "end_date": f"{year}-12-31",
        "daily": ",".join([
            "temperature_2m_mean",
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_sum"
        ]),
        "timezone": "auto",
        "models": "era5"
    }

    for attempt in range(3):

        try:

            response = requests.get(
                url,
                params=params,
                timeout=60
            )

            if response.status_code == 200:

                data = response.json()

                weather = pd.DataFrame(
                    data["daily"]
                )

                weather["Latitude"] = lat
                weather["Longitude"] = lon
                weather["Year"] = year

                weather.to_csv(
                    output_file,
                    index=False
                )

                return "SUCCESS"

            print(
                f"HTTP {response.status_code} "
                f"for {year}"
            )

        except Exception as e:

            print(
                f"Attempt {attempt + 1} failed "
                f"for {year}: {e}"
            )

            time.sleep(5)

    return "FAILED"
```

```python
import os
import time
import pandas as pd
import requests

WEATHER_DIR = "../data/weather"
os.makedirs(WEATHER_DIR, exist_ok=True)

CHECKPOINT_FILE = os.path.join(
    WEATHER_DIR,
    "weather_checkpoint.csv"
)

FAILED_FILE = os.path.join(
    WEATHER_DIR,
    "weather_failed.csv"
)
```

```python
if os.path.exists(CHECKPOINT_FILE):
    completed = pd.read_csv(CHECKPOINT_FILE)
else:
    completed = pd.DataFrame(
        columns=[
            "State",
            "District",
            "Year",
            "Latitude",
            "Longitude",
            "Status"
        ]
    )

print("Already completed:", len(completed))
```

```python
def fetch_weather_year(lat, lon, year):

    url = "https://archive-api.open-meteo.com/v1/archive"

    params = {
        "latitude": lat,
        "longitude": lon,
        "start_date": f"{year}-01-01",
        "end_date": f"{year}-12-31",
        "daily": ",".join([
            "temperature_2m_mean",
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_sum"
        ]),
        "timezone": "auto",
        "models": "era5"
    }

    for attempt in range(3):

        try:

            response = requests.get(
                url,
                params=params,
                timeout=60
            )

            if response.status_code == 200:

                data = response.json()

                weather = pd.DataFrame(
                    data["daily"]
                )

                weather["Latitude"] = lat
                weather["Longitude"] = lon
                weather["Year"] = year

                return weather

            print(
                f"HTTP {response.status_code}"
            )

        except Exception as e:

            print(
                f"Attempt {attempt + 1}: {e}"
            )

            time.sleep(5)

    return None
```

```python
failed_records = []

total = len(location_years)

for i, row in location_years.head(100).iterrows():

    state = row["State"]
    district = row["District"]
    year = int(row["Year"])

    # Check whether already completed
    already_done = completed[
        (completed["State"] == state) &
        (completed["District"] == district) &
        (completed["Year"] == year)
    ]

    if len(already_done) > 0:
        continue

    print(
        f"[{i + 1}/{total}] "
        f"{state} | {district} | {year}"
    )

    weather = fetch_weather_year(
        row["Latitude"],
        row["Longitude"],
        year
    )

    if weather is not None:

        # Save individual weather file
        filename = (
            f"{i}_{year}.csv"
        )

        filepath = os.path.join(
            WEATHER_DIR,
            filename
        )

        weather.to_csv(
            filepath,
            index=False
        )

        completed = pd.concat(
            [
                completed,
                pd.DataFrame([{
                    "State": state,
                    "District": district,
                    "Year": year,
                    "Latitude": row["Latitude"],
                    "Longitude": row["Longitude"],
                    "Status": "SUCCESS"
                }])
            ],
            ignore_index=True
        )

        completed.to_csv(
            CHECKPOINT_FILE,
            index=False
        )

    else:

        failed_records.append({
            "State": state,
            "District": district,
            "Year": year
        })

        pd.DataFrame(
            failed_records
        ).to_csv(
            FAILED_FILE,
            index=False
        )

    # Small delay
    time.sleep(0.5)

print("Download process finished.")
```

```python
def fetch_weather_batch(batch):

    latitudes = ",".join(
        batch["Latitude"].astype(str)
    )

    longitudes = ",".join(
        batch["Longitude"].astype(str)
    )

    url = "https://archive-api.open-meteo.com/v1/archive"

    params = {
        "latitude": latitudes,
        "longitude": longitudes,

        "start_date": "1997-01-01",
        "end_date": "2019-12-31",

        "daily": ",".join([
            "temperature_2m_mean",
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_sum"
        ]),

        "timezone": "auto",
        "models": "era5"
    }

    for attempt in range(3):

        try:

            response = requests.get(
                url,
                params=params,
                timeout=300
            )

            print(
                "HTTP status:",
                response.status_code
            )

            if response.status_code == 200:
                return response.json()

            if response.status_code == 429:
                print(
                    "Rate limited. Waiting 30 seconds..."
                )
                time.sleep(30)

            else:
                print(
                    response.text[:500]
                )

        except Exception as e:

            print(
                "Error:",
                e
            )

            time.sleep(10)

    return None
```

```python
import requests
import pandas as pd
import time

def fetch_weather_year(lat, lon, year):

    url = "https://archive-api.open-meteo.com/v1/archive"

    params = {
        "latitude": lat,
        "longitude": lon,
        "start_date": f"{year}-01-01",
        "end_date": f"{year}-12-31",
        "daily": ",".join([
            "temperature_2m_mean",
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_sum"
        ]),
        "timezone": "auto",
        "models": "era5"
    }

    for attempt in range(3):

        try:
            response = requests.get(
                url,
                params=params,
                timeout=60
            )

            if response.status_code == 200:

                data = response.json()

                weather = pd.DataFrame(
                    data["daily"]
                )

                return weather

            print(
                f"HTTP {response.status_code}"
            )

        except Exception as e:

            print(
                f"Attempt {attempt + 1} failed: {e}"
            )

            time.sleep(5)

    return None
```

```python
import unicodedata

def normalize_text(value):
    if pd.isna(value):
        return None

    value = str(value)

    value = unicodedata.normalize(
        "NFKD",
        value
    )

    value = "".join(
        c for c in value
        if not unicodedata.combining(c)
    )

    return (
        value
        .upper()
        .strip()
    )
```

```python
# ============================================
# RESOLVE THE 15 REMAINING CLIMATE RECORDS
# ============================================

# Calculate area directly in the same CRS
# as the downloaded ADM2 GeoJSON.
# We are using it only as a matching fingerprint,
# not as a real-world area measurement.

adm2_area = adm2.copy()

adm2_area["Geo_Area"] = (
    adm2_area.geometry.area
)

adm2_area["Climate_District_clean"] = (
    adm2_area["shapeName"]
    .astype(str)
    .apply(normalize_text)
)

# State from our spatial join
geo_state = adm2_with_state[
    [
        "shapeName__district",
        "shapeName__state"
    ]
].copy()

geo_state["Climate_District_clean"] = (
    geo_state["shapeName__district"]
    .astype(str)
    .apply(normalize_text)
)

geo_state["Climate_State_clean"] = (
    geo_state["shapeName__state"]
    .astype(str)
    .apply(normalize_text)
)

# Add state to ADM2 area table
adm2_area = adm2_area.merge(
    geo_state[
        [
            "Climate_District_clean",
            "Climate_State_clean"
        ]
    ].drop_duplicates(),
    on="Climate_District_clean",
    how="left"
)

# --------------------------------------------
# Get the 15 climate records still missing state
# --------------------------------------------

remaining_climate = climate_unique[
    climate_unique["Climate_State_clean"].isna()
].copy()

remaining_climate["Shape_Area"] = pd.to_numeric(
    remaining_climate["Shape_Area"],
    errors="coerce"
)

# --------------------------------------------
# Match by district name + closest polygon area
# --------------------------------------------

resolved_remaining = []

for _, row in remaining_climate.iterrows():

    candidates = adm2_area[
        adm2_area["Climate_District_clean"]
        == row["Climate_District_clean"]
    ].copy()

    if candidates.empty:
        print(
            "NO CANDIDATE:",
            row["shapeName"]
        )
        continue

    candidates["area_difference"] = (
        abs(
            candidates["Geo_Area"]
            - row["Shape_Area"]
        )
    )

    best = candidates.loc[
        candidates["area_difference"].idxmin()
    ]

    resolved_remaining.append({
        "shapeID": row["shapeID"],
        "shapeName": row["shapeName"],
        "Climate_State_clean": best[
            "Climate_State_clean"
        ],
        "area_difference": best[
            "area_difference"
        ]
    })

resolved_remaining = pd.DataFrame(
    resolved_remaining
)

print(
    resolved_remaining.to_string(index=False)
)

print(
    "\nResolved:",
    len(resolved_remaining)
)

print(
    "Maximum area difference:",
    resolved_remaining[
        "area_difference"
    ].max()
)
```

```python
# ============================================
# CORRECT POLYGON-LEVEL STATE + AREA REFERENCE
# ============================================

# Start from the original ADM2 polygons
polygon_ref = adm2.copy()

# Add the state found by the spatial join.
# IMPORTANT: use the row index, NOT district name.
polygon_ref["Climate_State_clean"] = (
    adm2_with_state["shapeName__state"]
    .apply(normalize_text)
)

polygon_ref["Climate_District_clean"] = (
    polygon_ref["shapeName"]
    .apply(normalize_text)
)

# Area in the same CRS used by the climate dataset
polygon_ref["Geo_Area"] = (
    polygon_ref.geometry.area
)

print(
    polygon_ref[
        [
            "shapeName",
            "Climate_State_clean",
            "Geo_Area"
        ]
    ].head(10).to_string(index=False)
)
```

```python
remaining_climate = climate_unique[
    climate_unique["Climate_State_clean"].isna()
].copy()

remaining_climate["Shape_Area"] = pd.to_numeric(
    remaining_climate["Shape_Area"],
    errors="coerce"
)

resolved_remaining = []

for _, row in remaining_climate.iterrows():

    candidates = polygon_ref[
        polygon_ref["Climate_District_clean"]
        == row["Climate_District_clean"]
    ].copy()

    if candidates.empty:
        print("NO CANDIDATE:", row["shapeName"])
        continue

    candidates["area_difference"] = (
        abs(
            candidates["Geo_Area"]
            - row["Shape_Area"]
        )
    )

    best = candidates.loc[
        candidates["area_difference"].idxmin()
    ]

    resolved_remaining.append({
        "shapeID": row["shapeID"],
        "shapeName": row["shapeName"],
        "Climate_State_clean": best["Climate_State_clean"],
        "area_difference": best["area_difference"]
    })

resolved_remaining = pd.DataFrame(
    resolved_remaining
)

print(
    resolved_remaining[
        [
            "shapeName",
            "Climate_State_clean",
            "area_difference"
        ]
    ].to_string(index=False)
)
```

```python
print(
    "\nResolved records:",
    len(resolved_remaining)
)

print(
    "\nMaximum area difference:",
    resolved_remaining["area_difference"].max()
)
```

```python
# ============================================
# BUILD FINAL CLIMATE STATE-DISTRICT DATA
# ============================================

# Add the resolved states to the 720 already-matched records
resolved_lookup = resolved_remaining[
    [
        "shapeID",
        "Climate_State_clean"
    ]
].copy()

climate_final = climate_unique.merge(
    resolved_lookup,
    on="shapeID",
    how="left",
    suffixes=("", "_resolved")
)

# Fill missing states using the polygon-level resolution
climate_final["Climate_State_clean"] = (
    climate_final["Climate_State_clean"]
    .fillna(
        climate_final["Climate_State_clean_resolved"]
    )
)

# Remove temporary column
climate_final = climate_final.drop(
    columns=["Climate_State_clean_resolved"]
)

print("Total climate records:", len(climate_final))

print(
    "Missing states:",
    climate_final["Climate_State_clean"].isna().sum()
)
```

```python
import numpy as np
import pandas as pd

model_df = final_dataset_v2[
    [
        "State",
        "District",
        "Crop",
        "Crop_Year",
        "Season",
        "Area",
        "Rainfall",
        "Temperature",
        "Yield"
    ]
].copy()

for col in ["State", "District", "Crop", "Season"]:
    model_df[col] = (
        model_df[col]
        .astype(str)
        .str.strip()
    )

for col in [
    "Crop_Year",
    "Area",
    "Rainfall",
    "Temperature",
    "Yield"
]:
    model_df[col] = pd.to_numeric(
        model_df[col],
        errors="coerce"
    )

model_df = model_df.dropna()

model_df = model_df[
    (model_df["Area"] > 0) &
    (model_df["Rainfall"] >= 0) &
    (model_df["Yield"] > 0)
].copy()

model_df["Log_Yield"] = np.log1p(
    model_df["Yield"]
)

print("Final model data:", model_df.shape)
```

```python
X = model_df[
    [
        "State",
        "District",
        "Crop",
        "Season",
        "Crop_Year",
        "Area",
        "Rainfall",
        "Temperature"
    ]
]

y = model_df["Log_Yield"]

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
from catboost import CatBoostRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

cat_features = [
    "State",
    "District",
    "Crop",
    "Season"
]

model = CatBoostRegressor(
    iterations=1000,
    learning_rate=0.05,
    depth=8,
    loss_function="RMSE",
    eval_metric="RMSE",
    random_seed=42,
    verbose=100
)

model.fit(
    X_train,
    y_train,
    cat_features=cat_features,
    eval_set=(X_test, y_test),
    early_stopping_rounds=100
)
```

```python
# ============================================
# CATBOOST EVALUATION
# ============================================

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predict log(Yield)
pred_log = model.predict(X_test)

# Convert back to original Yield
pred_yield = np.expm1(pred_log)
actual_yield = np.expm1(y_test)

# Metrics
mae = mean_absolute_error(
    actual_yield,
    pred_yield
)

rmse = np.sqrt(
    mean_squared_error(
        actual_yield,
        pred_yield
    )
)

r2 = r2_score(
    actual_yield,
    pred_yield
)

print("================================")
print("       CATBOOST RESULTS")
print("================================")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
```
